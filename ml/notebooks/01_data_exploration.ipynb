# WildTrace — 01. Data Exploration
Inspect the ATRW (Amur Tiger Re-identification in the Wild) dataset structure and image counts. Place ATRW under `ml/datasets/atrw/`.

```python
from pathlib import Path
from PIL import Image

atrw = Path('ml/datasets/atrw')
print('ATRW exists:', atrw.exists())
if atrw.exists():
    inds = sorted(p for p in atrw.iterdir() if p.is_dir())
    print(f'{len(inds)} individuals')
    for d in inds[:8]:
        imgs = list(d.glob('*.jpg')) + list(d.glob('*.png'))
        print(d.name, len(imgs), 'images')
```

```python
from PIL import Image
import matplotlib.pyplot as plt

# Preview a few sample tiger images
if atrw.exists():
    d = sorted(p for p in atrw.iterdir() if p.is_dir())[0]
    imgs = list(d.glob('*.jpg')) + list(d.glob('*.png'))
    fig, axs = plt.subplots(1, min(4, len(imgs)), figsize=(16, 4))
    for ax, ip in zip(axs, imgs[:4]):
        ax.imshow(Image.open(ip))
        ax.axis('off')
    plt.tight_layout()
    plt.show()
```
